```python
import os
import tkinter as tk
from tkinter import filedialog
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pydicom
from PIL import Image

import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, models, transforms

# ==========================================
# 1. Preprocessing & Dataset Setup
# ==========================================
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])

dataset_path = 'dataset'

if not os.path.exists(dataset_path):
  raise FileNotFoundError(
      f"❌ Directory '{dataset_path}' not found. Please create it and add"
      ' subfolders first.'
  )

dataset = datasets.ImageFolder(dataset_path, transform=transform)
loader = DataLoader(dataset, batch_size=16, shuffle=True)
class_names = dataset.classes
print(f'Detected Classes: {class_names}')

# ==========================================
# 2. Model Architecture & Training Loop
# ==========================================
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
model.fc = nn.Linear(model.fc.in_features, len(class_names))
model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.0001)

print('\n=== Starting Model Training ===')
model.train()
epochs = 5

for epoch in range(epochs):
  total_loss, correct = 0, 0
  for inputs, labels in loader:
    inputs, labels = inputs.to(device), labels.to(device)
    optimizer.zero_grad()
    outputs = model(inputs)
    loss = criterion(outputs, labels)
    loss.backward()
    optimizer.step()

    total_loss += loss.item()
    correct += (outputs.argmax(1) == labels).sum().item()

  print(
      f'Epoch {epoch+1}/{epochs} - Loss: {total_loss/len(loader):.4f} -'
      f' Accuracy: {correct/len(dataset):.2%}'
  )

# Save Checkpoint
checkpoint_path = 'brain_tumor_model.pth'
torch.save(
    {'model': model.state_dict(), 'classes': class_names}, checkpoint_path
)
print("✅ Model saved successfully to 'brain_tumor_model.pth'")


# ==========================================
# 3. DICOM and Standard Image Reader
# ==========================================
def read_image(path):
  ext = os.path.splitext(path)[1].lower()
  if ext in ['.dcm', '.dicom']:
    ds = pydicom.dcmread(path)
    img = ds.pixel_array.astype(float)
    img = (img - np.min(img)) / (np.max(img) - np.min(img) + 1e-5) * 255.0
    return Image.fromarray(img.astype(np.uint8)).convert('RGB')
  else:
    img_cv = cv2.imread(path)
    if img_cv is None:
      raise ValueError('Failed to read image file.')
    return Image.fromarray(cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB))


# ==========================================
# 4. Interactive File Picker & Inference
# ==========================================
print('\n=== Opening File Selection Dialog... ===')
root = tk.Tk()
root.withdraw()
root.attributes('-topmost', True)

file_path = filedialog.askopenfilename(
    title='Select Brain MRI Image',
    filetypes=[
        ('All Supported Formats', '*.dcm *.dicom *.png *.jpg *.jpeg *.tif *.bmp')
    ],
)

if file_path:
  model.eval()
  img = read_image(file_path)
  tensor = transform(img).unsqueeze(0).to(device)

  with torch.no_grad():
    probs = torch.softmax(model(tensor), dim=1)
    conf, pred_idx = probs.max(1)
    result = class_names[pred_idx.item()]
    conf_score = conf.item() * 100

  # Classification Logic
  if result == 'Not_Brain_MRI':
    status = f'REJECTED: Not a Brain MRI Image ({conf_score:.1f}%)'
    color = 'red'
  elif result == 'No_Tumor':
    status = (
        'ACCEPTED: Brain MRI | Diagnosis: Healthy / No Tumor Detected'
        f' ({conf_score:.1f}%)'
    )
    color = 'green'
  else:
    status = (
        'ACCEPTED: Brain MRI | Diagnosis: Brain Tumor Detected'
        f' ({conf_score:.1f}%)'
    )
    color = 'orange'

  print('\n[Final Diagnosis Result]:')
  print(status)

  # Display Output
  plt.figure(figsize=(7, 6))
  plt.imshow(img)
  plt.title(status, color=color, fontweight='bold', fontsize=10, pad=12)
  plt.axis('off')
  plt.tight_layout()
  plt.show()
else:
  print('❌ No file selected.')
```
